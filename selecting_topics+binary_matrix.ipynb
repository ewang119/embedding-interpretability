```python
train_docs=[doc_id for doc_id in reuters.fileids() if doc_id.startswith('training/')]

data= []
for doc_id in train_docs:
    data.append({
        'id': doc_id,
        'text': reuters.raw(doc_id),
        'topics': reuters.categories(doc_id)})
df= pd.DataFrame(data)
all_topics= reuters.categories()
all_topics
```

```python
topic_counts= df['topics'].explode().value_counts(ascending=True)
selected_topics =topic_counts.head(40).index.tolist()
print(f"40 Least Mentioned Topics:\n{selected_topics}\n")
```

```python
def filter_topics(topic_list):
    return [x for x in topic_list if x in selected_topics]

df['filtered_topics'] =df['topics'].apply(filter_topics)
mlb= MultiLabelBinarizer(classes=selected_topics)
binary_matrix= mlb.fit_transform(df['filtered_topics'])
labels_df =pd.DataFrame(binary_matrix, columns=mlb.classes_)
```
